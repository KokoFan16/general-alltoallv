# Plot benchmark results (gata / gatav)

Reads a benchmark output file from `gata_example` or `gatav_example`,
picks the best (minimum-time) configuration for tunable algorithms
(`MPICH-scattered` over `b`, `gata`/`gatav` over `(b, r)`), and:

- plots time vs. message size (log-log) with one curve per algorithm
- prints the best parameters for `MPICH-scattered` and `gata`/`gatav` at each `n`

Edit the `ATA_FILE` / `ATAV_FILE` paths below and run the cells.


```python
import re
from collections import defaultdict
import matplotlib.pyplot as plt

# Edit these paths to point at your benchmark output files.
ATA_FILE  = 'ata.txt'
ATAV_FILE = 'atav.txt'

# Matches:  [tag] nprocs, n, extra1, extra2, time
LINE_RE = re.compile(
    r'\[([^\]]+)\]\s+(\d+),\s+(\d+),\s+(\d+),\s+(\d+),\s+([\d.eE+-]+)'
)

ATA_TAGS  = ['MPI_Alltoall',  'OMPI-linear', 'OMPI-pairwise',
             'MPICH-scattered', 'Spreadout', 'Bruck', 'gata']
ATAV_TAGS = ['MPI_Alltoallv', 'OMPI-linear', 'OMPI-pairwise',
             'MPICH-scattered', 'Spreadout', 'Bruck', 'gatav']
```

```python
def parse(path):
    """Returns ({tag: {n: {(e1, e2): [time, ...]}}}, set_of_nprocs)."""
    data = defaultdict(lambda: defaultdict(lambda: defaultdict(list)))
    nprocs_seen = set()
    with open(path) as f:
        for line in f:
            m = LINE_RE.search(line)
            if not m:
                continue
            tag, nprocs, n, e1, e2, t = m.groups()
            nprocs_seen.add(int(nprocs))
            data[tag][int(n)][(int(e1), int(e2))].append(float(t))
    return data, nprocs_seen

def best_per_n(data_for_tag):
    """{n: ((e1, e2), best_time)}.  Best = min across all params and iterations."""
    result = {}
    for n, params in data_for_tag.items():
        best_p, best_t = None, float('inf')
        for p, times in params.items():
            t = min(times)
            if t < best_t:
                best_t, best_p = t, p
        result[n] = (best_p, best_t)
    return result
```

```python
def plot(data, variant, nprocs, save_to=None):
    tags = ATAV_TAGS if variant == 'atav' else ATA_TAGS
    ours = 'gatav' if variant == 'atav' else 'gata'
    title = ('Alltoallv (non-uniform)' if variant == 'atav'
             else 'Alltoall (uniform)') + f' — {nprocs} processes'

    fig, ax = plt.subplots(figsize=(11, 6.5))
    markers = ['o', 's', '^', 'v', 'D', 'P', '*', 'X']
    colors  = ['#888888', '#1f77b4', '#2ca02c', '#9467bd',
               '#8c564b', '#17becf', '#d62728']

    for i, tag in enumerate(tags):
        if tag not in data:
            continue
        best = best_per_n(data[tag])
        ns = sorted(best.keys())
        times = [best[n][1] for n in ns]
        is_ours = (tag == ours)
        ax.plot(ns, times,
                marker=markers[i % len(markers)],
                color=colors[i % len(colors)],
                label=tag,
                linewidth=3 if is_ours else 1.5,
                markersize=10 if is_ours else 6,
                zorder=10 if is_ours else 1)

    ax.set_xscale('log', base=2)
    ax.set_yscale('log')
    ax.set_xlabel('Message size per peer (elements, log\u2082)', fontsize=12)
    ax.set_ylabel('Time (seconds, log scale)', fontsize=12)
    ax.set_title(title, fontsize=13)
    ax.grid(True, which='both', linestyle='--', alpha=0.4)
    ax.legend(loc='best', fontsize=10)
    plt.tight_layout()
    if save_to:
        plt.savefig(save_to, dpi=150)
        print(f'Saved: {save_to}')
    plt.show()

def print_best_table(data, tag, param_names):
    """param_names = ('b',) or ('b', 'r')."""
    if tag not in data:
        print(f'(no data for {tag})')
        return
    cols = ['n'] + list(param_names) + ['time(s)']
    widths = [10] + [6] * len(param_names) + [14]
    header = '  '.join(f'{c:>{w}}' for c, w in zip(cols, widths))
    print(f'Best {tag} parameters per n:')
    print(header)
    print('-' * len(header))
    best = best_per_n(data[tag])
    for n in sorted(best.keys()):
        (e1, e2), t = best[n]
        if len(param_names) == 1:
            row_vals = [n, e1, t]
        else:
            row_vals = [n, e1, e2, t]
        row = '  '.join(
            (f'{v:>{w}}' if not isinstance(v, float) else f'{v:>{w}.6g}')
            for v, w in zip(row_vals, widths))
        print(row)
```

## ata (uniform alltoall)

```python
ata_data, ata_nprocs = parse(ATA_FILE)
nprocs = next(iter(ata_nprocs))
plot(ata_data, 'ata', nprocs, save_to='gata_comparison.png')
```

```python
print_best_table(ata_data, 'MPICH-scattered', ('b',))
print()
print_best_table(ata_data, 'gata', ('b', 'r'))
```

## atav (non-uniform alltoallv)

```python
atav_data, atav_nprocs = parse(ATAV_FILE)
nprocs = next(iter(atav_nprocs))
plot(atav_data, 'atav', nprocs, save_to='gatav_comparison.png')
```

```python
print_best_table(atav_data, 'MPICH-scattered', ('b',))
print()
print_best_table(atav_data, 'gatav', ('b', 'r'))
```
